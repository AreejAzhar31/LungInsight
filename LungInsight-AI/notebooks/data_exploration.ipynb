# LungInsight AI — Data Exploration
Quick sanity checks on the dataset before training: class balance, sample images, and image size distribution.

```python
import sys
sys.path.append('..')

from utils.config import load_config
from utils.dataset import ChestXrayDataset

cfg = load_config('../configs/config.yaml')
train_ds = ChestXrayDataset(cfg.data.train_dir, cfg.data.classes)
val_ds = ChestXrayDataset(cfg.data.val_dir, cfg.data.classes)
test_ds = ChestXrayDataset(cfg.data.test_dir, cfg.data.classes)

print('Train:', len(train_ds), train_ds.class_counts())
print('Val:  ', len(val_ds), val_ds.class_counts())
print('Test: ', len(test_ds), test_ds.class_counts())
```

```python
import matplotlib.pyplot as plt
import cv2

fig, axes = plt.subplots(1, 4, figsize=(14, 4))
for ax, cls in zip(axes, ['NORMAL', 'NORMAL', 'PNEUMONIA', 'PNEUMONIA']):
    samples = [s for s in train_ds.samples if train_ds.classes[s[1]] == cls]
    if not samples:
        continue
    path, _ = samples[0]
    img = cv2.cvtColor(cv2.imread(str(path)), cv2.COLOR_BGR2RGB)
    ax.imshow(img)
    ax.set_title(cls)
    ax.axis('off')
plt.tight_layout()
plt.show()
```

```python
# Class balance bar chart
import matplotlib.pyplot as plt

counts = train_ds.class_counts()
plt.bar(counts.keys(), counts.values(), color=['#4C72B0', '#DD8452'])
plt.title('Training Set Class Balance')
plt.ylabel('Image count')
plt.show()
```
